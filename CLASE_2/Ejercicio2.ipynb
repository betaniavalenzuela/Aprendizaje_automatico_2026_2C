### Imporatacion de librerias


```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

### Carga de archivo csv

```python
data_csv= pd.read_csv('Desktop/Automobile.csv', encoding= "ISO-8859-1")
```

```python
data_csv.head()
```

### Grafico de dispersion


```python
plt.figure(figsize=(8, 5))
plt.scatter(data_csv['weight'], data_csv['mpg'], alpha=0.7, color='tab:blue', edgecolors='k')

# Formato y títulos
plt.title('Relación entre el Peso del Vehículo y el Consumo (MPG)')
plt.xlabel('Peso (lbs)')
plt.ylabel('Millas por Galón (MPG)')
plt.grid(True, linestyle='--', alpha=0.5)

# Mostrar
plt.tight_layout()
plt.show()
```

### Grafico Histograma

```python

plt.figure(figsize=(8, 5))
plt.hist(data_csv['mpg'], bins=15, color='#4C72B0', edgecolor='black', alpha=0.8)

# etiquetas y títulos
plt.title('Distribución del Consumo de Combustible (MPG)', fontsize=14, fontweight='bold')
plt.xlabel('Millas por Galón (MPG)', fontsize=12)
plt.ylabel('Frecuencia (Cantidad de Vehículos)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Grafico Boxplot

```python
import seaborn as sns
```

```python
# Configurar la figura con dos subgráficos
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: MPG vs Cilindros
sns.boxplot(
    data=data_csv, 
    x='cylinders', 
    y='mpg', 
    hue='cylinders', 
    palette='Set2', 
    legend=False, 
    ax=axes[0]
)
axes[0].set_title('Consumo (MPG) según Número de Cilindros', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Cilindros', fontsize=11)
axes[0].set_ylabel('Millas por Galón (MPG)', fontsize=11)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Gráfico 2: MPG vs Origen
sns.boxplot(
    data=data_csv, 
    x='origin', 
    y='mpg', 
    hue='origin', 
    palette='Set2', 
    legend=False, 
    ax=axes[1]
)
axes[1].set_title('Consumo (MPG) según Origen del Vehículo', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Origen', fontsize=11)
axes[1].set_ylabel('Millas por Galón (MPG)', fontsize=11)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Grafico de Barras

```python

# Crear tabla de frecuencias cruzadas entre año y origen
frecuencias = pd.crosstab(data_csv['model_year'], data_csv['origin'])[['usa', 'japan', 'europe']]

# Graficar barras apiladas
ax = frecuencias.plot(kind='bar', stacked=True, figsize=(11, 5.5), colormap='Set2', edgecolor='black')

plt.title('Distribución Apilada de Automóviles por Año y Origen', fontsize=13, fontweight='bold')
plt.xlabel('Año del Modelo', fontsize=11)
plt.ylabel('Cantidad de Automóviles', fontsize=11)
plt.legend(title='Origen', labels=['Estados Unidos', 'Japón', 'Europa'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
