### Carga de archivo Excel

```python
data_excel = pd.read_excel('Desktop/inventario.xlsx', sheet_name='Inventario')
```

```python
data_excel.head()
```

```python
# Excluir la fila final de totales
df_productos = data_excel[data_excel['ID Producto'] != 'TOTALES / PROMEDIOS'].dropna(subset=['ID Producto'])

# Calcular el promedio de la columna 'Stock Actual'
promedio_stock = df_productos['Stock Actual'].mean()

print(f"El promedio de stock disponible es: {promedio_stock:.2f} unidades")
```
